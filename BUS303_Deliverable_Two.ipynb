```python
columns_to_check = [
    'Total Fat (g)',
    'Saturated Fat (g)',
    'Trans Fat (g)',
    'Cholesterol (mg)',
    'Sodium (mg)',
    'Fiber (g)',
    'Protein (g)',
    'Carbs (g)',
    'Sugars (g)'
]

for column in columns_to_check:
    print(f"\n{column}:")
    print(dnkn[column].unique())
```

```python
dnkn1 = dnkn.copy()
dnkn1
```

```python
food_categories = [
    'Donuts', 'Sandwiches', 'Muffins', 'Kolache',
    'Soft Serve', 'Hash Browns', 'Dunkin’ Bowls'
]

dnkn2 = dnkn1[~dnkn1['Category'].isin(food_categories)].copy()

dnkn2['Category'].value_counts()
```

```python
dnkn2 = dnkn2.drop_duplicates()

print("Duplicate rows remaining:", dnkn2.duplicated().sum())
print("Total rows:", len(dnkn2))
dnkn2
```

```python
dnkn2 = dnkn2.replace('\xa0', '')
dnkn2 = dnkn2.replace('<5', '')

dnkn2
```

```python
print("Non-breaking spaces remaining:", (dnkn2 == '\xa0').sum().sum())
print("<5 values remaining:", (dnkn2 == '<5').sum().sum())
```

```python
import numpy as np

dnkn2 = dnkn2.replace('', np.nan)

dnkn2.isnull().sum()
```

```python
print("Missing values by column:")
print(dnkn2.isnull().sum())

print("Total rows with missing values:", dnkn2.isnull().any(axis=1).sum())
```

```python
print(dnkn['Cholesterol (mg)'].value_counts().head(15))
```

```python
dnkn3 = dnkn2.dropna().copy()

print("Final dataset shape:", dnkn3.shape)
dnkn3
```

```python
print("Missing values remaining:", dnkn3.isnull().sum().sum())
print("Dataset shape:", dnkn3.shape)
```

```python
numeric_columns = dnkn3.columns.drop(['Category', 'Item'])

dnkn3[numeric_columns] = dnkn3[numeric_columns].apply(
    pd.to_numeric
)

dnkn3.dtypes
```

```python
print("Dataset shape:", dnkn3.shape)
print("Missing values:", dnkn3.isnull().sum().sum())
print("Duplicate rows:", dnkn3.duplicated().sum())
print("\nData types:")
print(dnkn3.dtypes)
```

```python
dnkn3.to_csv("BUS303_Dunkin_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
